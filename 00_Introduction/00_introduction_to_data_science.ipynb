## 🔬 What is Data Science?

**Data Science** is the field of using **data, programming, statistics, and analytical techniques** to extract useful insights, discover patterns, and support better decision-making.

### 🔹 In Simple Words

> **Data Science = Collecting + Cleaning + Analyzing + Understanding Data**

Data Scientists use data to answer questions such as:

* 📊 Descrptive Analysis - What happened?
* 🔍 Diagnostic Analysis - Why did it happen?
* 📈 Predictive Analysis - What might happen next?
* 💡 Prescriptive Analysis - What action should we take?

### 🧩 Key Components of Data Science

| Component                 | Purpose                                 |
| ------------------------- | --------------------------------------- |
| 🐍 **Programming**        | Work with and process data              |
| 📊 **Statistics**         | Understand patterns and relationships   |
| 🧹 **Data Cleaning**      | Prepare data for analysis               |
| 📈 **Data Visualization** | Communicate insights clearly            |
| 🤖 **Machine Learning**   | Make predictions and automate decisions |
| 🧠 **Domain Knowledge**   | Understand the real-world problem       |

### 💡 Example

An online shopping company can use Data Science to analyze customer data and:

* Understand what customers buy
* Find purchasing patterns
* Recommend products
* Predict future sales
* Detect unusual transactions

### 🎯 Goal of Data Science

The main goal of Data Science is to **turn raw data into useful insights and knowledge that can help solve real-world problems and make better decisions.**


### Why Data Science

We live in a world that runs on data. Everytime you pay through UPI, watch a video on youtube, browse through Amazon or Flipkart, order food etc you are creating data. Companies want to understand the data, it helps them in making better decesion. So we turn this data into useful answers.

### What is Data

Raw Facts, Figures or details about something is called Data. Example (Your Age, The city you live in, Price of products etc)

### 📊 Types of Data

#### - Based on a What Data contains
**Data Type**|Made of What
:--|:--
**Numeric Data**|Made of Numbers (eg - Marks scored, Age etc)
**Text Data**|Made of Letters and Words (eg - Name, Address etc)
**Image Data**|Pictures and photots (eg - selfie etc)
**Audio Data**|Sound Recordings (eg - Voice note, Bollywood song etc)
**Video Data**|Moving pictures with sound (eg Youtube video etc)

#### - Based on How Data is organised
**Data Type**|How it Organised
:--|:--
**Structured Data**|Neat organised data that fits perfectly into **Rows and Columns** (Tables). Eg- A Class marksheet which has Name, Rollno., Marks etc for each student.
**Semi Structured Data**|Data that has some structure but not a perfect table. Eg- JSON, Data from webste.
**Unstructured Data**|Data with no fic=xed format at all. It is the hardest type to work with. Eg- Photos, Videos.

### 📚 What is Information?

**Information** is meaningful and processed data that helps us **understand something, make decisions, or gain knowledge**.

### 🔹 Data vs Information

* **Data** → Raw facts and values
* **Information** → Data that has been processed or organized to give meaning

### 💡 Example

Suppose we have the following data:

`85, 72, 90, 65, 88`

These are just **raw values**.

After analyzing them, we can say:

> **The average score is 80%.**

Now the data has been processed and given meaning, so it becomes **information**.

### 🧠 Simple Definition

> **Information = Processed data that has meaning and is useful.**

### 🔄 Data Science Workflow

The **Data Science Workflow** is a series of steps used to solve a problem using data.

### 🧭 Typical Data Science Workflow

```text
🎯 Define the Problem
        ↓
📥 Collect Data
        ↓
🧹 Clean & Prepare Data
        ↓
🔍 Explore Data (EDA)
        ↓
📊 Analyze & Visualize Data
        ↓
🤖 Build a Model (if required)
        ↓
📈 Evaluate Results
        ↓
💡 Communicate Insights
        ↓
🚀 Deploy / Take Action
```

### 📌 Steps Explained

| Step                        | What happens?                                                            |
| --------------------------- | ------------------------------------------------------------------------ |
| 🎯 **Define the Problem**   | Understand the question or business problem                              |
| 📥 **Collect Data**         | Gather relevant data from databases, files, APIs, surveys, etc.          |
| 🧹 **Clean & Prepare Data** | Handle missing values, duplicates, incorrect data, and formatting issues |
| 🔍 **Explore Data**         | Find patterns, relationships, trends, and unusual values                 |
| 📊 **Analyze & Visualize**  | Use statistics and charts to understand the data                         |
| 🤖 **Build a Model**        | Use Machine Learning when prediction or classification is needed         |
| 📈 **Evaluate Results**     | Check how well the analysis or model performs                            |
| 💡 **Communicate Insights** | Present findings using reports, visualizations, or dashboards            |
| 🚀 **Deploy / Take Action** | Put the solution into use and make decisions based on the results        |

> 💡 **Note:** Not every Data Science project requires Machine Learning. Many projects focus only on **data analysis, visualization, and extracting insights**.

### 🔍 Data Science vs Data Analyst

| 📊 Data Analyst                    | 🔬 Data Scientist                          |
| ---------------------------------- | ------------------------------------------ |
| Analyzes existing data             | Analyzes data and builds predictive models |
| Focuses on **what happened & why** | Focuses on **what might happen**           |
| SQL, Excel, Python, Power BI       | Python, SQL, Statistics, Machine Learning  |
| Reports, dashboards, insights      | Models, predictions, advanced insights     |


> 💡 **In simple words:**
> **Data Analyst → Understands data**
> **Data Scientist → Uses data to predict and solve problems**
