## 🐼 Introduction to Pandas

**Pandas** is a Python library used for **working with data tables, data manipulation and analysis**.

It provides easy-to-use data structures such as **Series** and **DataFrame** for working with structured data.
* Series - A single column of data with a label (eg - Maths marks of a class).
* DataFrame - A full table with many column (eg - Marks of all subjects in separate columns of a class).

### 🔹 What can Pandas do?

* 📥 Read data from CSV, Excel, SQL, etc.
* 🔍 Select and filter data
* 🧹 Handle missing and duplicate values
* 📊 Analyze and summarize data
* 🔄 Sort, group, and combine data
* 💾 Save processed data

> 💡 **In simple words:**
> **Pandas helps us work with, clean, and analyze data efficiently in Python.**

### 📊 Creating a Series and DataFrame

Pandas mainly provides two important data structures:

* **Series** → One-dimensional data
* **DataFrame** → Two-dimensional tabular data

#### 🔹 Creating a Series

A Series can be created from a list with a label called an index (by default starts from 0):

```python
import pandas as pd
#default Index
marks = pd.Series([85, 90, 78, 92])
# We can also give our own custom index
# marks = pd.Series([85, 90, 78, 92],index=["Maths","Science","English","Computer"])

print(marks)
# We can also fetch any value by its index
# print(marks[0])
# print(marks["Maths"])
```

Every column in a DataFrame is actually a series.

#### 🔹 Creating a DataFrame

A DataFrame can be created using a dictionary(most common) or from a csv file(most important):

1. When created using a Dictionary each key becomes a column name and each list becomes its values.

```python
data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [20, 21, 19],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

2. 📥 Reading a **CSV (Comma-Separated Values)** File

Pandas provides the `read_csv()` function to read CSV files into a **DataFrame**.

##### 🔹 Example

```python
import pandas as pd
# If the file is present in the same folder then just specifying the name works otherwise we have to specify the whole path.
df = pd.read_csv("data.csv") 

# df = pd.read_csv("data/data.csv")

print(df)
```

##### 🔹 Display the First Few Rows

```python
df.head()
```

##### 🔹 Display the Last Few Rows

```python
df.tail()
```

> 💡 **Remember:** `pd.read_csv()` → reads a CSV file and returns a **DataFrame**.

### 🎯 Selecting Data in Pandas

Pandas provides different ways to select **columns, rows, and specific values** from a DataFrame.

#### 🔹 Selecting a Single Column

```python
df["Name"]
```

This returns a **Series**.

#### 🔹 Selecting Multiple Columns

```python
df[["Name", "Age"]]
```

This returns a **DataFrame**.

#### 🔹 Selecting Rows with `iloc`

`iloc` selects data using **position numbers**.

```python
df.iloc[0]      # First row
df.iloc[0:3]    # First three rows
```

#### 🔹 Selecting Rows with `loc`

`loc` selects data using **labels or index names**.

```python
df.loc[0]
```

#### 🔹 Selecting Specific Rows and Columns

```python
df.loc[0:2, ["Name", "Age"]]
```

```python
df.iloc[0:3, 0:2]
```

##### `.idxmax()`
Note: this is a **pandas** method (not NumPy) — there's no `np.idmax()`. 
It returns the **index label** of the row with the maximum value in a 
Series, which is what you'd normally use with a DataFrame.
```python
label = df["Percentage"].idxmax()
print(label)
```

#### 🧠 Remember

```text
loc   → Select using labels
iloc  → Select using positions

["Column"]        → Single column
[["Col1", "Col2"]] → Multiple columns
```

> 💡 `loc` and `iloc` are two of the most commonly used methods for selecting data in Pandas.

### 🔍 Filtering Data in Pandas

Filtering allows us to select rows that match a specific condition.

#### 🔹 Example Data

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 17, 22, 19],
    "Marks": [85, 72, 90, 65]
}

df = pd.DataFrame(data)
```

#### 🔹 Filter Using One Condition

Select students who are 18 or older:

```python
df[df["Age"] >= 18]
```

Select students with marks greater than 80:

```python
df[df["Marks"] > 80]
```

#### 🔹 Filter Using Multiple Conditions

##### AND (`&`)

```python
df[(df["Age"] >= 18) & (df["Marks"] > 80)]
```

Both conditions must be `True`.

##### OR (`|`)

```python
df[(df["Age"] >= 18) | (df["Marks"] > 80)]
```

At least one condition must be `True`.

##### NOT (`~`)

```python
df[~(df["Age"] >= 18)]
```

Selects rows where the condition is `False`.

##### 🧠 Remember

```text
&  → AND
|  → OR
~  → NOT
```

> 💡 When filtering with multiple conditions in Pandas, use `&` and `|` instead of Python's `and` and `or`.

### 🔃 Sorting Data in Pandas

Pandas allows us to sort a DataFrame by one or more columns using `sort_values()`.

#### 🔹 Sort by One Column

Sort by marks in ascending order:

```python
df.sort_values("Marks")
```

Sort by marks in descending order:

```python
df.sort_values("Marks", ascending=False)
```

#### 🔹 Sort by Multiple Columns

```python
df.sort_values(["Age", "Marks"])
```

This first sorts by **Age**, then by **Marks** when the ages are the same.

#### 🔹 Save the Sorted Data

```python
df = df.sort_values("Marks", ascending=False)
```

##### 🧠 Remember

```text
ascending=True   → Small → Large
ascending=False  → Large → Small
```

> 💡 `sort_values()` returns a sorted DataFrame without changing the original DataFrame unless you assign the result back to `df`.

### 👥 GroupBy in Pandas

`groupby()` is used to **group rows based on a column** and then perform calculations on each group.

#### 🔹 Example Data

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Department": ["IT", "HR", "IT", "HR"],
    "Salary": [50000, 45000, 60000, 55000]
}

df = pd.DataFrame(data)
```

#### 🔹 Group by a Column

Find the average salary of each department:

```python
df.groupby("Department")["Salary"].mean()
```

#### 🔹 Common Functions

```python
df.groupby("Department")["Salary"].sum()
df.groupby("Department")["Salary"].mean()
df.groupby("Department")["Salary"].max()
df.groupby("Department")["Salary"].min()
df.groupby("Department")["Salary"].count()
```

#### 🧠 Remember

```text
groupby()
   ↓
Group rows
   ↓
Apply a calculation
   ↓
Get summarized results
```

> 💡 **GroupBy = Split → Apply → Combine**

### ➕ Adding New Columns

We can create a new column by assigning values, using existing columns, or applying a function.

#### 🔹 Add a Column with Fixed Values

```python
df["City"] = ["Delhi", "Mumbai", "Pune", "Jaipur"]
```

#### 🔹 Add a Column Using Existing Columns

```python
df["Total"] = df["Maths"] + df["Science"]
```

#### 🔹 Add a Column Using `apply()`

`apply()` allows us to apply a function to each value in a column.

```python
def grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    else:
        return "C"

df["Grade"] = df["Marks"].apply(grade)
```

### 🧠 Remember

```text
df["New_Column"] = ...
        ↓
Create a new column

.apply(function)
        ↓
Apply a function to each value
```

> 💡 `apply()` is especially useful when creating a new column based on custom logic.

### ✏️ Updating Existing Values

We can update existing values in a DataFrame by selecting the **column, row, or condition** we want to change.

#### 🔹 Update a Single Value

```python
df.loc[0, "Marks"] = 95
```

This changes the marks of the first row to `95`.

#### 🔹 Update Multiple Values Using a Condition

```python
df.loc[df["Marks"] < 40, "Marks"] = 40
```

This changes all marks below `40` to `40`.

### 🔹 Update an Entire Column

```python
df["Marks"] = df["Marks"] + 5
```

This adds `5` to every value in the `Marks` column.



> 💡 Updating values is useful when correcting, transforming, or modifying existing data.

### 🗑️ Removing Data

We can remove **columns or rows** from a DataFrame using the `drop()` method.

#### 🔹 Remove a Column

```python
df = df.drop("Age", axis=1)
```

`axis=1` means we are removing a **column**.

#### 🔹 Remove Multiple Columns

```python
df = df.drop(["Age", "City"], axis=1)
```

#### 🔹 Remove a Row

```python
df = df.drop(0)
```

This removes the row with index `0`.

#### 🔹 Remove Multiple Rows

```python
df = df.drop([0, 2])
```

##### 🧠 Remember

```text
axis=0 → Rows
axis=1 → Columns
```

> 💡 `drop()` returns a new DataFrame. Assign it back to `df` if you want the change to remain.

### 💾 Saving Data

After modifying or analyzing a DataFrame, we can save it to a file using Pandas.

#### 🔹 Save as CSV

```python
df.to_csv("output.csv", index=False)
```

`index=False` prevents Pandas from saving the DataFrame index as an extra column.

#### 🔹 Save as Excel

```python
df.to_excel("output.xlsx", index=False)
```

#### 🔹 Save to a Specific Folder

```python
df.to_csv("data/output.csv", index=False)
```

##### 🧠 Remember

```text
read_csv()  → CSV → DataFrame
to_csv()    → DataFrame → CSV

read_excel() → Excel → DataFrame
to_excel()   → DataFrame → Excel
```

> 💡 Use `to_csv()` when you want to save your processed DataFrame as a CSV file.
