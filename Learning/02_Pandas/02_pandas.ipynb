## 🧹 Data Cleaning and Preprocessing

### 🔧 What is Data Cleaning?

**Data Cleaning** is the process of identifying and fixing (or removing) 
incorrect, incomplete, duplicate, or inconsistent data within a dataset. 
Real-world data is often messy — it may contain missing values, 
duplicate rows, wrong data types, or inconsistent formatting — and 
cleaning it ensures accurate and reliable analysis.

**Why it matters:**
- Prevents incorrect conclusions from bad data
- Improves the accuracy of any analysis or model built on the data
- Ensures consistency across the dataset (e.g., same date format, no duplicate entries)

**Common data cleaning tasks:**
- Handling missing values (NaN)
- Removing duplicate rows
- Fixing incorrect data types
- Standardizing text/categorical values (e.g., "male", "Male", "M" → one format)
- Detecting and handling outliers
- Wrong Data Types
- Renaming unclear column names

### 🔍 Exploring Dirty Data

Before cleaning a dataset, we first need to explore it to understand 
what issues exist — missing values, duplicates, inconsistent text, or 
wrong data types. Below are the common methods used for this, along 
with their code.

##### `.info()`
Gives a quick summary of the dataframe — column names, number of 
non-null values, and data types of each column.
```python
df.info()
```

##### `.isnull()`
Returns a dataframe of the same shape with `True`/`False` values, 
marking where data is missing (NaN) and where it isn't.
```python
df.isnull()
```

##### `.isnull().sum()`
Counts the number of missing values in each column — more useful 
than `.isnull()` alone since it summarizes rather than showing every cell.
```python
df.isnull().sum()
```

##### `.notnull()`
The opposite of `.isnull()` — returns `True` where data exists and 
`False` where it's missing.
```python
df.notnull()
```

##### `.duplicated()`
Returns `True`/`False` for each row, marking whether it's a duplicate 
of a previous row.
```python
df.duplicated()
```

##### `.duplicated().sum()`
Counts the total number of duplicate rows in the dataframe.
```python
df.duplicated().sum()
```

##### `.describe()`
Generates summary statistics (count, mean, std, min, max, quartiles) 
for numeric columns — useful for spotting outliers or invalid values.
```python
df.describe()
```

##### `.unique()`
Returns all distinct values in a column — helpful for spotting 
inconsistent text formatting (e.g., "Male", "male", "M").
```python
df["Gender"].unique()
```

##### `.value_counts()`
Counts how many times each unique value appears in a column — useful 
for spotting inconsistent categories at a glance (e.g., seeing "Male" 
30 times and "male" 5 times tells you they should be merged).
```python
df["Gender"].value_counts()
```

##### `.dtypes`
Shows the data type of every column — helps catch columns stored 
with the wrong type (e.g., numbers stored as text).
```python
df.dtypes
```

##### `.shape`
Returns the number of rows and columns as a tuple `(rows, columns)`.
```python
df.shape
```

### 🩹 Handling Missing Values

Once we've identified missing values using `.isnull().sum()`, the 
next step is deciding how to handle them — either by removing them 
or filling them in with a suitable value. Below are the common 
methods used, along with their code.

##### `.dropna()`
Removes rows (or columns) that contain missing values. Useful when 
missing data is rare and dropping a few rows won't significantly 
affect the analysis.
```python
df.dropna()
```

##### `.dropna(subset=[...])`
Drops rows only if specific columns have missing values, instead of 
checking the entire row.
```python
df.dropna(subset=["Attendance"])
```

##### `.fillna(value)`
Fills missing values with a specified constant value.
```python
df["Attendance"] = df["Attendance"].fillna(0)
```

##### `.fillna(mean)`
Fills missing numeric values with the column's average — a common 
choice since it doesn't drastically shift the data's distribution.
```python
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].mean())
```

##### `.fillna(median)`
Fills missing numeric values with the median — a better choice than 
mean when the column has outliers, since median isn't affected by 
extreme values.
```python
df["Percentage"] = df["Percentage"].fillna(df["Percentage"].median())
```

##### `.fillna(mode)`
Fills missing categorical values with the most frequently occurring 
value in that column.
```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
```

##### `.fillna(method="ffill")`
Forward-fill — fills a missing value using the value from the 
previous row. Common in time-series data.
```python
df["Age"] = df["Age"].fillna(method="ffill")
```

##### `.fillna(method="bfill")`
Backward-fill — fills a missing value

### 🗑️ Removing Duplicates

Duplicate rows can distort analysis by over-counting certain records. 
Below are the common methods used to detect and remove them.

##### `.drop_duplicates()`
Removes duplicate rows, keeping only the first occurrence by default.
```python
df = df.drop_duplicates()
```

##### `.drop_duplicates(subset=[...])`
Removes duplicates based on specific columns only, instead of 
checking every column.
```python
df = df.drop_duplicates(subset=["Name", "ID"])
```

##### `.drop_duplicates(keep="last")`
Keeps the last occurrence of a duplicate instead of the first.
```python
df = df.drop_duplicates(keep="last")
```

## 🔤 Cleaning Text Data

Text/categorical columns often have inconsistent formatting (extra 
spaces, mismatched capitalization, typos) that need to be standardized 
before analysis.

##### `.str.lower()`
Converts all text in a column to lowercase, useful for standardizing 
values like "Male", "male", "MALE" into one consistent format.
```python
df["Gender"] = df["Gender"].str.lower()
```

##### `.str.upper()`
Converts all text in a column to uppercase.
```python
df["City"] = df["City"].str.upper()
```

##### `.str.strip()`
Removes leading and trailing whitespace from text values.
```python
df["City"] = df["City"].str.strip()
```

##### `.str.title()`
Converts text to title case (first letter of each word capitalized) — 
useful for names and city values.
```python
df["City"] = df["City"].str.title()
```

##### `.str.replace()`
Replaces specific text/characters within a column — useful for fixing 
typos or standardizing inconsistent labels.
```python
df["Gender"] = df["Gender"].str.replace("Female", "F")
```

##### `.replace()`
Replaces specific values across a column (or dataframe) with new 
values — useful for mapping multiple inconsistent labels to one.
```python
df["Gender"] = df["Gender"].replace({"male": "M", "Female": "F"})
```

### 🏷️ Rename Column

Column names are sometimes unclear, inconsistent, or contain extra 
spaces/symbols — renaming them makes the dataset easier to work with.

##### `.rename(columns={...})`
Renames one or more specific columns using a dictionary mapping old 
names to new names.
```python
df = df.rename(columns={"Percentage": "Percent", "Attendance": "Attend"})
```

##### `.columns = [...]`
Renames all columns at once by assigning a new list of names — the 
list must match the total number of columns.
```python
df.columns = ["ID", "Name", "Gender", "Age", "City", "Class",
              "Maths", "Science", "English", "Attendance",
              "Percentage", "Status"]
```

##### `.columns.str.lower()`
Converts all column names to lowercase — useful for keeping naming 
consistent across a dataset.
```python
df.columns = df.columns.str.lower()
```

### 🔄 Change Data Types

Columns sometimes get stored with the wrong data type (e.g., numbers 
as text), which can cause errors during calculations or analysis.

##### `.astype()`
Converts a column to a specified data type.
```python
df["Age"] = df["Age"].astype(int)
```

##### `pd.to_numeric()`
Converts a column to a numeric type, and can handle invalid values 
gracefully using the `errors` parameter.
```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
```
errors="coerce" tells pandas: if a value can't be converted, don't crash — just turn it into NaN instead.

##### `pd.to_datetime()`
Converts a column to datetime format — useful for date/time columns 
stored as text.
```python
df["Date"] = pd.to_datetime(df["Date"])
```

##### `.dtypes`
Checks the current data type of every column, useful for verifying 
that conversions were applied correctly.
```python
df.dtypes
```

## 📅 Cleaning Date Columns

Date columns are often stored as plain text and can have inconsistent 
formats (e.g., "12-05-2024", "2024/05/12", "May 12, 2024"), which 
makes them hard to sort, filter, or calculate with. Converting them 
into a proper datetime format fixes this.

##### `pd.to_datetime()`
Converts a column of text/strings into pandas' datetime format, 
allowing date-based operations like sorting and filtering.
```python
df["Date"] = pd.to_datetime(df["Date"])
```

##### `pd.to_datetime(errors="coerce")`
Converts to datetime, but turns any unparseable values into `NaT` 
(Not a Time — the datetime equivalent of NaN) instead of raising an error.
```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
```

##### `pd.to_datetime(format=...)`
Specifies the exact format of the date strings — useful (and faster) 
when the format is known and consistent, or when auto-detection 
guesses wrong.
```python
df["Date"] = pd.to_datetime(df["Date"], format="%d-%m-%Y")
```

##### `.dt.year`
Extracts just the year from a datetime column.
```python
df["Year"] = df["Date"].dt.year
```

##### `.dt.month`
Extracts just the month from a datetime column.
```python
df["Month"] = df["Date"].dt.month
```

##### `.dt.day`
Extracts just the day from a datetime column.
```python
df["Day"] = df["Date"].dt.day
```

##### `.dt.day_name()`
Extracts the name of the weekday (e.g., "Monday") from a datetime column.
```python
df["Weekday"] = df["Date"].dt.day_name()
```

##### `.isnull().sum()` (on the date column)
After conversion, checks how many dates failed to parse and became `NaT`.
```python
df["Date"].isnull().sum()
```

### 🔧 Replacing Incorrect Values

Datasets often contain values that are technically valid data types 
but logically incorrect — like an age of 150, a negative attendance, 
or a percentage above 100. These need to be identified and replaced 
with more sensible values.

##### `.replace()`
Replaces a specific incorrect value with a new one across a column.
```python
df["Age"] = df["Age"].replace(150, 16)
```

##### Conditional replacement using `.loc[]`
Finds rows matching a condition and replaces the value in a specific 
column for just those rows — more flexible than `.replace()` when 
you don't know the exact incorrect value in advance.
```python
df.loc[df["Attendance"] < 0, "Attendance"] = 0
```

##### Capping values using `.clip()`
Restricts a column's values to a defined minimum/maximum range — 
useful for fixing out-of-range values like a percentage above 100 
or attendance above 100.
```python
df["Percentage"] = df["Percentage"].clip(lower=0, upper=100)
```

##### Replacing with the column mean/median
Replaces an identified incorrect value with a statistically sensible 
one, similar to handling missing values.
```python
df.loc[df["Age"] > 100, "Age"] = df["Age"].median()
```

##### Using `np.where()` for conditional replacement
Replaces values based on a condition, keeping the original value 
where the condition is False.
```python
import numpy as np
df["Attendance"] = np.where(df["Attendance"] < 0, 0, df["Attendance"])
```
Negative values changed to zero.

##### `.apply()` with a custom function
Applies custom logic to replace values that don't fit a simple 
condition — useful for more complex fixing rules.
```python
def fix_age(age):
    if age > 100 or age < 5:
        return None
    return age

df["Age"] = df["Age"].apply(fix_age)
```

### 🗂️ Data Cleaning Workflow

1. **Explore the Data** — understand structure and spot issues
   `.info()`, `.shape`, `.describe()`, `.isnull().sum()`, 
   `.duplicated().sum()`, `.unique()`, `.value_counts()`, `.dtypes`

2. **Handle Missing Values** — decide to drop or fill
   `.dropna()`, `.fillna()`, `.interpolate()`

3. **Remove Duplicates** — eliminate repeated rows
   `.duplicated()`, `.drop_duplicates()`

4. **Clean Text Data** — standardize inconsistent formatting
   `.str.lower()`, `.str.strip()`, `.str.title()`, `.replace()`

5. **Rename Columns** — make column names clear and consistent
   `.rename()`, `.columns`

6. **Fix Data Types** — convert columns to correct types
   `.astype()`, `pd.to_numeric()`, `pd.to_datetime()`

7. **Clean Date Columns** — parse and extract date components
   `pd.to_datetime()`, `.dt.year`, `.dt.month`, `.dt.day_name()`

8. **Replace Incorrect/Invalid Values** — fix logically wrong data
   `.replace()`, `.loc[]` conditional fixes, `.clip()`, `np.where()`

9. **Final Check** — confirm the dataset is clean
   `.info()`, `.isnull().sum()`, `.duplicated().sum()`, `.describe()`
