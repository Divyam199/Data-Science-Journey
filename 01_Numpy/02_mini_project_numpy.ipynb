## 👩🏼‍🦰Student Marks Analysis

#### Let us create a table which contains following details about students and answer real questions about them using Numpy.
    1. Names
    2. Marks 
    3. Attendance
    4. Percentage


---

#### 🔢Impoting Numpy

```python
import numpy as np # Importing numpy Library
```

---

#### 🗄️Creating arrays

```python

first_names = [
    "Aarav", "Vivaan", "Aditya", "Vihaan", "Arjun", "Sai", "Reyansh", "Krishna",
    "Ishaan", "Rohan", "Ananya", "Diya", "Isha", "Kavya", "Myra", "Anika",
    "Saanvi", "Aadhya", "Riya", "Zara", "Kabir", "Dhruv", "Aryan", "Yash",
    "Om", "Shaurya", "Neha", "Priya", "Tanya", "Meera", "Nisha", "Pooja",
    "Rahul", "Karan", "Aman", "Varun", "Nikhil", "Siddharth", "Rajat", "Manav",
    "Simran", "Kiara", "Aarohi", "Ira", "Navya", "Pihu", "Anaya", "Kiaan",
    "Advait", "Reyan"
]

np.random.seed(42)  # remove for different random data each run
names = np.array(first_names)

marks = np.random.randint(35, 100, size=50)          # marks out of 100
attendance = np.random.randint(60, 100, size=50)      # attendance %
percentage = np.round((marks / 100) * 100, 2)         # percentage scored
print("Names : ", names)
print("Marks : ", marks)
print("Attendance : ", attendance)
print("Percentage : ",percentage)
```

---

#### Q. Find Names of those students who scored Highest Marks, Lowest Marks❓

```python
# Finding the Highest and Lowest marks
highest_score = np.max(marks)  
lowest_score = np.min(marks)

# Creating a string of the Names who scored Highest marks
highest_names = list(names[marks == highest_score])

if len(highest_names)>1:
    highest_names = ",".join(highest_names[:-1])+" and "+highest_names[-1]

else:
    highest_names = highest_names[0]

# Creating a string of the Names who scored Lowest marks
lowest_names = list(names[marks == lowest_score])

if len(lowest_names)>1:
    lowest_names = ", ".join(lowest_names[:-1])+" and "+lowest_names[-1]

else:
    lowest_names = lowest_names[0]

print(f"Highest Marks scored by {highest_names} : {highest_score}")
print(f"Lowest Marks scored by {lowest_names} : {lowest_score}")
```

---

#### Q. Find Names of those students who scored above 80 and those who scored below 40❓

```python
# Finding the marks above 80 and below 40
above_80 = marks[marks>80]
below_35 = marks[marks<40]

# Finding the Names for those who scored above 80
above_80_names = names[marks > 80]

# Finding the Names for those who scored below 40
below_35_names = names[marks < 40]

print("Students who scored above 80")
for name, mark in zip(above_80_names, above_80):
    print(name," : ",mark) 

print("\nStudents who scored below 40")
for name, mark in zip(below_35_names, below_35):
    print(name," : ",mark) 
```

---

#### Q. Find and display the top 3 and bottom 3 students with there marks and attendace ❓

```python
# Sorting the percentage array to get top 3 and bottom 3
percent_indices_sorted = np.argsort(percentage)
percent_top_3 = marks[percent_indices_sorted[-3:]][::-1]
percent_bottom_3 = marks[percent_indices_sorted[:3]]

# Finding the top 3 and bottom 3 names of the students
top_3_names = names[percent_indices_sorted[-3:]][::-1]
bottom_3_names = names[percent_indices_sorted[:3]]

# Finding top 3 and Bottom 3 attendance for the students
attendance_top_3 = attendance[percent_indices_sorted[-3:]][::-1]
attendance_bottom_3 = attendance[percent_indices_sorted[:3]]


print("Top 3 Students")
print(f"{"Name":<12} {"Percentage":>12} {"Attendance":>12}")  
for name, percent, attendances in zip(top_3_names, percent_top_3, attendance_top_3):
    print(f"{name:<12}  {percent:>12}  {attendances:>12}") 

print("\nBottom 3 Students")
print(f"{"Name":<12} {"Percentage":>12} {"Attendance":>12}") 
for name, percent, attendances in zip(bottom_3_names, percent_bottom_3, attendance_bottom_3):
    print(f"{name:<12}  {percent:>12}  {attendances:>12}")

# It is highly likely that the top 3 students will have higher percentage than the bottom 3 students but inour example the attendance is generated randomly 😄.
```

---

#### Q. Find and display the average of the class but without using np.mean() ❓

```python
# Getting the total marks for the class
total_marks = np.sum(marks)
total_students = len(names)

# Calculating the average
average = total_marks/total_students
print("Average of the class is : ", average)
```
